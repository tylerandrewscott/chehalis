# GPT-4o Mini EDS Form Extractor

Extract structured data from Executive Document Summary (EDS) forms using GPT-4o Mini vision capabilities.

## Key Features:
1. **Vision-based extraction**: Can detect checkboxes, marks, and visual elements
2. **Single-pass extraction**: Gets all fields in one API call
3. **Cost-effective**: ~$0.002 per form (~$80 for 40k forms)
4. **Testing Mode**: Quick validation on example forms
5. **Progress Tracking**: Visual progress bar with ETA
6. **Resume Capability**: Automatically skip processed files

## Quick Start:
- **Testing:** Set `TESTING_MODE = True`, add PDFs to `_exampleforms/`
- **Production:** Set `TESTING_MODE = False`

## Define Output Schema

````python
# Schema definition for GPT-4o Mini to follow
OUTPUT_SCHEMA = {
    "eds_number": "string",
    "date_prepared": "YYYY-MM-DD format",
    "agency": {
        "name": "string",
        "contact_person": "string",
        "telephone": "string",
        "email": "string"
    },
    "contract_info": {
        "professional_personal_services": "boolean (true if checkbox marked)",
        "grant": "boolean",
        "lease": "boolean",
        "attorney": "boolean",
        "mou": "boolean",
        "qpa": "boolean",
        "contract_for_procured_services": "boolean",
        "maintenance": "boolean",
        "license_agreement": "boolean",
        "amendment_number": "string or null",
        "renewal_number": "string or null",
        "other": "boolean",
        "other_details": "string or null"
    },
    "vendor": {
        "id": "string",
        "name": "string",
        "telephone": "string",
        "email": "string",
        "address": "string",
        "registered_with_sos": "boolean or null",
        "minority_owned": "boolean or null",
        "women_owned": "boolean or null"
    },
    "sub_vendor": {
        "minority_owned": "boolean or null",
        "women_owned": "boolean or null"
    },
    "fiscal": {
        "account_number": "string",
        "account_name": "string",
        "amount_this_action": "number or null",
        "new_contract_total": "number or null",
        "revenue_generated_this_action": "number or null",
        "revenue_generated_total": "number or null"
    },
    "time_period": {
        "from_date": "YYYY-MM-DD or null",
        "to_date": "YYYY-MM-DD or null"
    },
    "source_selection": {
        "bid_quotation": "boolean",
        "rfp_number": "string or null",
        "emergency": "boolean",
        "negotiated": "boolean",
        "special_procurement": "boolean",
        "other": "boolean",
        "other_details": "string or null"
    },
    "additional_info": {
        "renewal_language": "boolean or null",
        "termination_for_convenience": "boolean or null"
    }
}

print("✓ Output schema defined")
````

## Helper Functions

````python
def get_lowest_page_number(form_pages_str: str) -> Optional[int]:
    """Extract the lowest page number from form_pages string."""
    if pd.isna(form_pages_str) or form_pages_str == "":
        return None
    
    if ',' in str(form_pages_str):
        page_numbers = [int(x.strip()) for x in str(form_pages_str).split(',')]
    else:
        page_numbers = [int(str(form_pages_str).strip())]
    
    return min(page_numbers)


def pdf_page_to_image(pdf_path: str, page_number: int, dpi: int = DPI) -> Image.Image:
    """Convert a single PDF page to PIL Image."""
    doc = fitz.open(pdf_path)
    page = doc.load_page(page_number - 1)
    
    mat = fitz.Matrix(dpi / 72, dpi / 72)
    pix = page.get_pixmap(matrix=mat)
    img_data = pix.tobytes("png")
    
    doc.close()
    
    img = Image.open(BytesIO(img_data))
    
    # Resize if too large
    if max(img.size) > MAX_IMAGE_SIZE:
        ratio = MAX_IMAGE_SIZE / max(img.size)
        new_size = tuple(int(dim * ratio) for dim in img.size)
        img = img.resize(new_size, Image.Resampling.LANCZOS)
    
    return img


def image_to_base64(image: Image.Image) -> str:
    """Convert PIL Image to base64 string."""
    buffered = BytesIO()
    image.save(buffered, format="PNG")
    return base64.b64encode(buffered.getvalue()).decode('utf-8')


def validate_json_output(json_path: Path) -> bool:
    """Validate that saved JSON is well-formed and contains required fields."""
    try:
        with open(json_path, 'r') as f:
            data = json.load(f)
        
        if 'error' in data:
            return True
        
        if 'structured_data' not in data:
            return False
        
        return True
        
    except json.JSONDecodeError:
        return False
    except Exception:
        return False


print("✓ Helper functions loaded")
````

## GPT-4o Mini Extraction Function

````python
def extract_with_gpt4o_mini(image: Image.Image, filename: str = "") -> Dict[str, Any]:
    """
    Extract structured data from EDS form using GPT-4o Mini vision.
    """
    results = {
        'processing_timestamp': datetime.now().isoformat(),
        'model_name': MODEL_NAME,
        'dpi': DPI,
        'testing_mode': TESTING_MODE,
        'structured_data': {},
    }
    
    try:
        # Convert image to base64
        base64_image = image_to_base64(image)
        
        # Create prompt
        prompt = f"""You are extracting data from an Executive Document Summary (EDS) form.

CRITICAL INSTRUCTIONS:
1. Look carefully at underscored blank areas to the left of response options - an option is marked if it has an X or checkmark above the underscore.
2. For yes/no questions, look at which option (Yes or No) has a mark (X or checkmark) above the underscored blank area to the left of each option.
3. Extract ALL text exactly as written
4. For dollar amounts, extract as numbers without $ or commas (e.g., 150000 not $150,000)
5. For dates, use YYYY-MM-DD format
6. Return ONLY valid JSON matching the schema below, no other text

OUTPUT SCHEMA:
{json.dumps(OUTPUT_SCHEMA, indent=2)}

FORM FIELD LOCATIONS:
- Section 1: EDS Number (top area)
- Section 2: Date Prepared
- Section 3: Contract Types (CONTRACTS & LEASES) - look for checkboxes with X or marks
- Section 4-9: Fiscal information (Account Number, Account Name, Amounts)
- Section 11-12: Time period (From/To dates)
- Section 13: Method of source selection - look for checkboxes with X or marks
- Section 14: Agency Name
- Section 23-27: Vendor information
- Section 28: Vendor registered with SOS (Yes/No - look which is marked)
- Section 29: Primary Vendor M/WBE - Minority and Women (Yes/No for each)
- Section 31: Sub Vendor M/WBE - Minority and Women (Yes/No for each)
- Section 33: Renewal Language (Yes/No)
- Section 34: Termination for Convenience (Yes/No)
- Agency Contact Information section: Contact person, telephone, email

Extract all data from this form and return as JSON:"""
        
        # Call GPT-4o Mini
        response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=[
                {
                    "role": "user",
                    "content": [
                        {
                            "type": "text",
                            "text": prompt
                        },
                        {
                            "type": "image_url",
                            "image_url": {
                                "url": f"data:image/png;base64,{base64_image}",
                                "detail": "high"
                            }
                        }
                    ]
                }
            ],
            max_tokens=2000,
            temperature=0
        )
        
        # Extract JSON from response
        response_text = response.choices[0].message.content
        
        # Try to parse JSON (handle markdown code blocks)
        json_match = re.search(r'```json\s*(.+?)\s*```', response_text, re.DOTALL)
        if json_match:
            json_text = json_match.group(1)
        else:
            # Try to find JSON object directly
            json_match = re.search(r'\{.+\}', response_text, re.DOTALL)
            if json_match:
                json_text = json_match.group(0)
            else:
                json_text = response_text
        
        structured_data = json.loads(json_text)
        results['structured_data'] = structured_data
        
        # Store token usage for cost tracking
        results['token_usage'] = {
            'prompt_tokens': response.usage.prompt_tokens,
            'completion_tokens': response.usage.completion_tokens,
            'total_tokens': response.usage.total_tokens
        }
        
        # Calculate cost (GPT-4o Mini pricing)
        # Input: $0.150 per 1M tokens, Output: $0.600 per 1M tokens
        input_cost = (response.usage.prompt_tokens / 1_000_000) * 0.150
        output_cost = (response.usage.completion_tokens / 1_000_000) * 0.600
        results['cost_usd'] = input_cost + output_cost
        
    except json.JSONDecodeError as e:
        results['error'] = f"JSON parsing error: {str(e)}"
        results['raw_response'] = response_text
    except Exception as e:
        results['error'] = str(e)
    
    return results


print("✓ Extraction function loaded")
````

## Load and Filter Data

````python
if TESTING_MODE:
    print("\n=== TESTING MODE: Loading Example Forms ===")
    
    # Find all PDF files in the testing directory
    testing_path = Path(TESTING_DIR)
    pdf_files = list(testing_path.glob("*.pdf"))
    
    if not pdf_files:
        print(f"⚠️  No PDF files found in {TESTING_DIR}")
        print("Please add example PDF files to the testing directory.")
    else:
        print(f"Found {len(pdf_files)} PDF files")
    
    # Create a simple dataframe for testing
    forms_df = pd.DataFrame([
        {
            'filename': pdf_file.name,
            'full_path': str(pdf_file),
            'form_pages': '1',  # Assume first page for testing
            'num_form_pages': 1,
            'lowest_page': 1
        }
        for pdf_file in pdf_files
    ])
    
    # Apply sample size limit if specified
    if TESTING_SAMPLE_SIZE and len(forms_df) > TESTING_SAMPLE_SIZE:
        print(f"Limiting to first {TESTING_SAMPLE_SIZE} files for quick testing")
        forms_df = forms_df.head(TESTING_SAMPLE_SIZE)
    
    print(f"Prepared {len(forms_df)} documents for testing")
    
else:
    print("\n=== PRODUCTION MODE: Loading from CSV ===")
    
    # Load professional services contracts data
    with open("../../../data/raw/indiana_prof_services_contracts.json", 'r') as f:
        prof_services_contracts = json.load(f)
    
    print(f"Total professional services contracts: {len(prof_services_contracts)}")
    
    # Apply agency filter if specified
    if FILTER_AGENCIES is not None:
        print(f"=== AGENCY FILTER APPLIED ===")
        target_agencies = [FILTER_AGENCIES] if isinstance(FILTER_AGENCIES, str) else FILTER_AGENCIES
        print(f"Filtering for agencies: {target_agencies}")
        
        original_count = len(prof_services_contracts)
        prof_services_contracts = [
            contract for contract in prof_services_contracts 
            if contract['agencyName'] in target_agencies
        ]
        print(f"Contracts after filter: {len(prof_services_contracts)} (from {original_count})")
    
    # Extract filenames
    prof_services_filenames = set()
    for contract in prof_services_contracts:
        filename = contract['pdfUrl'].split('/')[-1]
        prof_services_filenames.add(filename)
    
    print(f"Unique professional services PDF files: {len(prof_services_filenames)}")
    
    # Load CSV and filter for professional services with forms
    df = pd.read_csv(CSV_PATH, low_memory=False)
    print(f"Total documents in CSV: {len(df)}")
    
    forms_df = df[df['contains_form'] == True].copy()
    print(f"Documents with forms: {len(forms_df)}")
    
    forms_df = forms_df[forms_df['filename'].isin(prof_services_filenames)].copy()
    print(f"Professional services documents with forms: {len(forms_df)}")
    
    # Add lowest page number
    forms_df['lowest_page'] = forms_df['form_pages'].apply(get_lowest_page_number)
    forms_df = forms_df.dropna(subset=['lowest_page'])
    forms_df['lowest_page'] = forms_df['lowest_page'].astype(int)
    print(f"Documents with valid page numbers: {len(forms_df)}")
````

## Check Processing Status

````python
print(f"\n{'='*50}")
print("CHECKPOINT STATUS")
print(f"{'='*50}")

if TESTING_MODE:
    print("🧪 MODE: TESTING")
    print(f"Source: {TESTING_DIR}")
else:
    print("🚀 MODE: PRODUCTION")
    if FILTER_AGENCIES is not None:
        if isinstance(FILTER_AGENCIES, str):
            print(f"🔍 Filter: {FILTER_AGENCIES}")
        else:
            print(f"🔍 Filter: {', '.join(FILTER_AGENCIES)}")

print(f"Total documents queued: {len(forms_df)}")
print(f"DPI: {DPI}")
print(f"Model: {MODEL_NAME}")

# Check existing files
already_processed = 0
for idx, row in forms_df.iterrows():
    filename_base = Path(row['filename']).stem
    json_filename = f"{filename_base}_page_{row['lowest_page']}_gpt4o.json"
    json_path = output_dir / json_filename
    if json_path.exists() and json_path.stat().st_size > 0:
        already_processed += 1

remaining = len(forms_df) - already_processed
print(f"Already completed: {already_processed} files")
print(f"Remaining to process: {remaining} files")
print(f"Progress: {(already_processed/len(forms_df)*100):.1f}% complete")

if remaining > 0:
    estimated_cost = remaining * 0.002  # $0.002 per form estimate
    print(f"\n💰 Estimated cost for remaining: ${estimated_cost:.2f}")

if not CLOBBER and already_processed > 0:
    print("\n💡 TIP: Will skip already processed files (CLOBBER=False)")
    print("💡 To reprocess: Set CLOBBER=True")

print(f"{'='*50}\n")
````

## Process Documents

````python
# Initialize counters
processed_count = 0
skipped_count = 0
error_count = 0
validation_failures = 0
total_cost = 0.0

print("Starting document processing...\n")

# Create progress bar
progress_bar = tqdm(
    forms_df.iterrows(),
    total=len(forms_df),
    desc="Processing forms",
    unit="file"
)

for idx, row in progress_bar:
    filename = row['filename']
    lowest_page = row['lowest_page']
    
    # Update progress bar description
    progress_bar.set_postfix_str(f"{filename[:30]}...")
    
    # Generate output filename
    filename_base = Path(filename).stem
    json_filename = f"{filename_base}_page_{lowest_page}_gpt4o.json"
    json_path = output_dir / json_filename
    
    # Check if already processed
    if not CLOBBER and json_path.exists() and json_path.stat().st_size > 0:
        skipped_count += 1
        continue
    
    # Construct PDF path
    if TESTING_MODE:
        pdf_path = Path(row['full_path'])
    else:
        pdf_path = Path(CONTRACTS_DIR) / filename
    
    if not pdf_path.exists():
        print(f"\nPDF not found: {filename}")
        error_count += 1
        continue
    
    try:
        # Convert PDF page to image
        image = pdf_page_to_image(str(pdf_path), lowest_page, dpi=DPI)
        
        # Extract with GPT-4o Mini
        results = extract_with_gpt4o_mini(image, filename)
        
        # Add metadata
        results['source_file'] = filename
        results['source_page'] = lowest_page
        
        # Track cost
        if 'cost_usd' in results:
            total_cost += results['cost_usd']
        
        # Save results
        with open(json_path, 'w') as f:
            json.dump(results, f, indent=2)
        
        # Validate saved JSON
        if not validate_json_output(json_path):
            validation_failures += 1
        
        processed_count += 1
        
        # Log summary if verbose
        if VERBOSE and 'structured_data' in results:
            sd = results['structured_data']
            print(f"\n  EDS: {sd.get('eds_number', 'N/A')} | "
                  f"Agency: {sd.get('agency', {}).get('name', 'N/A')} | "
                  f"Cost: ${results.get('cost_usd', 0):.4f}")
        
    except Exception as e:
        print(f"\nError processing {filename}: {str(e)}")
        error_count += 1
        
        # Save error info
        error_data = {
            'source_file': filename,
            'source_page': lowest_page,
            'error': str(e),
            'processing_timestamp': datetime.now().isoformat(),
            'testing_mode': TESTING_MODE
        }
        with open(json_path, 'w') as f:
            json.dump(error_data, f, indent=2)

# Close progress bar
progress_bar.close()

# Final summary
print(f"\n{'='*50}")
print("PROCESSING COMPLETE")
print(f"{'='*50}")
print(f"✓ Processed: {processed_count}")
print(f"⊘ Skipped (already done): {skipped_count}")
print(f"✗ Errors: {error_count}")
if validation_failures > 0:
    print(f"⚠ Validation failures: {validation_failures}")
print(f"Total: {processed_count + skipped_count + error_count}")
print(f"\n💰 Total cost: ${total_cost:.2f}")
print(f"💰 Average cost per form: ${total_cost/processed_count:.4f}" if processed_count > 0 else "")

if TESTING_MODE:
    print(f"\n📁 Results: {TESTING_OUTPUT_DIR}")
else:
    print(f"\n📁 Results: {OUTPUT_DIR}")

print(f"{'='*50}")
````

## Create Summary Statistics

````python
# Analyze all processed files
json_files = list(output_dir.glob("*_gpt4o.json"))

if json_files:
    print(f"\n{'='*50}")
    print("SUMMARY STATISTICS")
    print(f"{'='*50}")
    print(f"Mode: {'TESTING' if TESTING_MODE else 'PRODUCTION'}")
    print(f"Total files: {len(json_files)}")
    
    # Collect statistics
    stats = {
        'total_files': len(json_files),
        'files_with_errors': 0,
        'contract_types': {},
        'source_selection_methods': {},
        'minority_owned': 0,
        'women_owned': 0,
        'total_amount': 0.0,
        'files_with_amount': 0,
        'total_cost': 0.0
    }
    
    for json_file in json_files:
        with open(json_file, 'r') as f:
            data = json.load(f)
        
        # Track cost
        if 'cost_usd' in data:
            stats['total_cost'] += data['cost_usd']
        
        if 'error' in data:
            stats['files_with_errors'] += 1
            continue
        
        if 'structured_data' not in data:
            continue
        
        sd = data['structured_data']
        
        # Contract types
        if 'contract_info' in sd:
            for key, value in sd['contract_info'].items():
                if value is True:
                    stats['contract_types'][key] = stats['contract_types'].get(key, 0) + 1
        
        # Source selection
        if 'source_selection' in sd:
            for key, value in sd['source_selection'].items():
                if value is True and not key.endswith('_details'):
                    stats['source_selection_methods'][key] = stats['source_selection_methods'].get(key, 0) + 1
        
        # Vendor diversity
        if sd.get('vendor', {}).get('minority_owned'):
            stats['minority_owned'] += 1
        if sd.get('vendor', {}).get('women_owned'):
            stats['women_owned'] += 1
        
        # Amounts
        amount = sd.get('fiscal', {}).get('amount_this_action')
        if amount:
            stats['total_amount'] += amount
            stats['files_with_amount'] += 1
    
    print(f"\nFiles with errors: {stats['files_with_errors']}")
    print(f"💰 Total cost: ${stats['total_cost']:.2f}")
    print(f"💰 Average cost per file: ${stats['total_cost']/len(json_files):.4f}")
    
    if stats['contract_types']:
        print(f"\nContract Types (top 5):")
        for ctype, count in sorted(stats['contract_types'].items(), key=lambda x: x[1], reverse=True)[:5]:
            print(f"  {ctype}: {count}")
    
    if stats['source_selection_methods']:
        print(f"\nSource Selection Methods:")
        for method, count in sorted(stats['source_selection_methods'].items(), key=lambda x: x[1], reverse=True):
            print(f"  {method}: {count}")
    
    print(f"\nVendor Diversity:")
    print(f"  Minority-owned: {stats['minority_owned']}")
    print(f"  Women-owned: {stats['women_owned']}")
    
    print(f"\nFiscal Information:")
    print(f"  Files with amount data: {stats['files_with_amount']}")
    if stats['files_with_amount'] > 0:
        print(f"  Total contract amount: ${stats['total_amount']:,.2f}")
        print(f"  Average contract amount: ${stats['total_amount']/stats['files_with_amount']:,.2f}")
    
    print(f"{'='*50}")
else:
    print("No processed files found.")
````
